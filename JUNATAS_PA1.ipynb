```python
def rotate_word(text): # Define Function
    return text[1:] + text[0] #Code that puts first index to the end 

print(f"2ECEA - {rotate_word('2ECEA')}") #Examples to show how the code works
print(f"Computer - {rotate_word('Computer')}")
print(f"Keyboard - {rotate_word('Keyboard')}")
print(f"Mouse - {rotate_word('Mouse')}")
print(f"Junatas - {rotate_word('Junatas')}")

user = input("\nEnter word to rotate: ")
r = rotate_word(user)
print(f"Orginal: {user} - Rotated: {r}")
```

```python
def make_username(first_name, last_name): # Define Function
    firstname_changed = first_name.lower().replace(" ", "") # Turns the first name to lower case
    lastname_changed = last_name.lower().replace(" ", "") # Turns the last name to lower case and replace space with dot
    return firstname_changed + "." + lastname_changed

print(f"Elijah, Junatas - {make_username('Elijah', 'Junatas')}") #Examples to show how the code works
print(f"James, Yap - {make_username('James', 'Yap')}")
print(f"Cardo, Dalisay - {make_username('Cardo', 'Dalisay')}")

first = input("\nInput First Name: ")
last = input("Input Last Name: ")
print(make_username(first, last))
```

```python
def swap_bookends(items): # Define Function #Swaps the first and last elements of a list using sequence unpacking.
    first, *middle, last = items #Unpack the list using extended sequence unpacking
    return [last] + middle + [first #Create a new list with swapped ends

print(f"1,2,3,4 - {swap_bookends([1,2,3,4])}") #Examples to show how the code works
print(f"pink, blue, black - {swap_bookends(['pink', 'blue', 'black'])}")
print(f"11, 7 - {swap_bookends([11, 7])}")   

user = input("\nEnter your list: ")
items = [item.strip() for item in user.split(",")]
print(swap_bookends(items))
```
